# Графики

```python
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import plotly.io as pio

custom_style = go.layout.Template(pio.templates["plotly_white"])

custom_style.layout.update(
    paper_bgcolor="#fcfbfa",
    plot_bgcolor="#f7f5f0",
    font=dict(
        family="Segoe UI, Arial, sans-serif",
        size=13,
        color="#3d4047"
    ),

    margin=dict(t=80, b=60, l=60, r=40),
    title=dict(
        font=dict(size=22, color="#1c212d", weight="bold"),
        pad=dict(b=10)
    )
)

custom_style.layout.colorway = [
    "#deb887",
    "#e28743",
    "#6b8e6b",
    "#e06d75",
    "#205072"
]

axis_config = dict(
    gridcolor="#eae6df",
    zerolinecolor="#d3cbd0",
    tickcolor="#c8becc",
    ticklen=6,
    linecolor="#eae6df",
    showgrid=True,
    ticks="outside",
    title=dict(font=dict(weight="bold", color="#1c212d")),
    tickfont=dict(
        size=12,
        color="#545861",
        family="Arial, sans-serif"
    )
)

custom_style.layout.xaxis = go.layout.XAxis(axis_config)
custom_style.layout.yaxis = go.layout.YAxis(axis_config)

custom_style.layout.legend = dict(
    bgcolor="rgba(252, 251, 250, 0.9)",
    bordercolor="#eae6df",
    borderwidth=1,
    orientation="h",
    yanchor="bottom", y=1.02,
    xanchor="right", x=1
)


pio.templates["summer_minimal"] = custom_style
pio.templates.default = "summer_minimal"
```

```python

# Сортируем данные по времени старта
df_sorted = combined_metrics.sort_values('datetime_start').reset_index(drop=True)

# Добавляем колонку-порядковый номер (ось X)
df_sorted['order'] = df_sorted.index + 1  # чтобы начиналось с 1

# Строим точечный график без линий
fig = px.scatter(
    df_sorted,
    x='order',
    y='value',
    color='method',
    title='F1-score Кросс-Валидация на 5 фолдах',
    labels={
        'order': 'Номер эксперимента',
        'value': 'F1',
        'method': 'Метод'
    }
)

fig.update_traces(marker=dict(size=18))

fig.show()
```

# Регрессии

```python
import pandas as pd
names = ['lin_reg_l1.csv', 'lin_reg_l2.csv', 'rf_reg.csv',
         'svmreg_results.csv', 'catboost_reg_results.csv']
# df = pd.read_csv
def get_df(name):
    df = pd.read_csv(name)
    df['method'] = name.split('.')[0]
    return df.sample(10)
dfs = [get_df(name) for name in names]
```

```python
combined_metrics = pd.concat(dfs, axis=0)
combined_metrics
```

```python

# Сортируем данные по времени старта
df_sorted = combined_metrics.sort_values('datetime_start').reset_index(drop=True)

# Добавляем колонку-порядковый номер (ось X)
df_sorted['order'] = df_sorted.index + 1  # чтобы начиналось с 1
df_sorted['value'] *= -1
# Строим точечный график без линий
fig = px.scatter(
    df_sorted,
    x='order',
    y='value',
    color='method',
    title='MAE-score Кросс-Валидация на 5 фолдах',
    labels={
        'order': 'Номер эксперимента',
        'value': 'MAE',
        'method': 'Метод'
    }
)

fig.update_traces(marker=dict(size=18))

fig.show()
```
