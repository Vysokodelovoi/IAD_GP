## Оценивание и штрафы



Групповой проект оценивается исходя из устной защиты (презентации) командой выполненного ими проекта перед комиссией. Комиссия составляется из преподавателей курса, ассистентов курса, а также приглашенных экспертов из индустрии.

В рамках каждой защиты комиссия выставляет два типа оценок.

Первая оценка — это оценка для всего проекта в целом. Она ставится комиссией, исходя из глубины проработки командой предметной области, сложности применяемых в проекте инструментов, а также прочих аспектов, связанных с содержанием выполняемого в проекте ТЗ. При выставлении этой оценки комиссия ориентируется на критерии, более подробно представленные ниже в разделе "Критерии оценки проекта".

Вторая оценка — это индивидуальная оценка каждого участника команды. Именно эта оценка выставляется впоследствии в ведомость. Индивидуальная оценка участника команды определяется комиссией, исходя из того, насколько участник ориентируется в выполненном командой проекте, как он отвечает на адресованные персонально ему вопросы комиссии, насколько в принципе прослеживается его роль в проекте, а также исходя из других аспектов, связанных с участием конкретного человека в деятельности команды. Индивидуальная оценка не может превосходить оценку команды за проект в целом (первую оценку), однако, в случае когда участник спокойно и корректно отвечает на все адресованные ему вопросы, свободно ориентируется в материале и содержании выполненного проекта, а также когда его роль в работе команды не вызывает сомнений, — в таком случае, индивидуальная оценка участника становится равной оценке для всего проекта (т.е. в идеальном случае — вторая оценка равна первой оценке). Кроме того, в исключительных случаях, при демонстрации участником на защите превосходящих ожидания уровней понимания, участия и вовлечения в рамках проекта, комиссия может своим решением повысить индивидуальную оценку для этого участника на 0,5 или 1 балл относительно общей оценки за проект (первой оценки).

Во избежание субъективности и/или предвзятости оценки тех или иных членов комиссии в отношении тех или иных команд, а также каких-либо иных проявлений неравных условий — распределение команд между комиссиями осуществляется абсолютно рандомно с использованием средств библиотеки Numpy :)) — в качестве рэндом сида выступает дата, в которую производится распределение, например, `np.random.seed(29052026)`.

Обратите, пожалуйста, внимание на то, что бОльшая часть комиссий осуществляет приём проектов в очной форме, однако могут присутствовать и 1-2 комиссии, которые будут работать дистанционно с использованием средств видеосвязи. В случае попадания команды путем случайного распределения в комиссию, работающую в очной форме, *не допускается* какое-либо изменение формата проведения для команды на онлайн-режим, равно как и любые другие перемещения команд между комиссиями (смены комиссий), для сохранения честности и непредвзятости случайного распределения.

# Часть 1. HotelFlow: прогнозирование отмен бронирований и оценка ценности бронирований

## Введение и постановка задачи

### Описание бизнес-контекста

В данном проекте рассматривается  онлайн-платформа бронирования отелей **HotelFlow**. По своей логике HotelFlow похожа на сервисы типа Booking.com. Платформа агрегирует данные о бронированиях, отелях, клиентах, каналах продаж, датах заезда, длительности проживания, стоимости проживания и отменах.

Для такой платформы важно заранее понимать, какие бронирования могут быть отменены, а какие являются наиболее ценными с точки зрения потенциальной выручки. Отмена бронирования, в свою очередь, может приводить к потере дохода, особенно, если клиент отменит бронь незадолго до даты заезда и отель не успеет повторно продать номер.

Также стоит отметить, что не все бронирования одинаково ценны для компании. Некоторые клиенты потенциально приносят больше выручки, например, так как бронируют более дорогие номера. Если платформа модет заранее выделять такие бронирования, она будет способна приоритизировать работу с ними, например, предлагать дополнительные услуги, персональные предложения или специальные условия для удержания клиента.

Проект HotelFlow направлен на построение моделей машинного обучения, которые помогают платформе принимать более обоснованные бизнес-решения на основе исторических данных о бронированиях.

В рамках проекта мы будем рассматривать две связанные бизнес-проблемы риск отмены бронирования и оценку ожидаемой стоимости бронирования. Совместное использование этих оценок позволит платформе не только понимать вероятность отмены, но и оценивать потенциальную ценность каждого бронирования для бизнеса.

### Цель проекта

Наша цель проекта - построить систему анализа бронирований для онлайн-платформы HotelFlow, которая решает две задачи:

1. заранее предсказывать вероятность отмены бронирования
2. определять ценность бронирования

### Постановка бизнес-задач

#### Задача 1: Предсказание отмены бронирования



Первая бизнес-задача заключается в определении, отменит ли клиент бронирование.

Если платформа будет заранее понимать, что определённое бронирование имеет высокий риск отмены, она может использовать эту информацию для принятия решений. Например, можно отправить клиенту напоминание, предложить подтвердить бронирование, применить более осторожную стратегию overbooking или заранее скорректировать прогноз загрузки отеля.

Польза для компании:
* отель заранее видит рискованные бронирования
* точнее планировать загрузку номеров
* аккуратнее использовать overbooking
* отправлять напоминания клиентам с высоким риском отмены
* снижать потери от пустующих номеров
* лучше оценивать будущую выручку

#### Задача 2: Определение ценности бронирования

Вторая бизнес-задача заключается в прогнозировании `adr` - ожидаемой средней стоимости проживания за ночь.

Для HotelFlow важно понимать потенциальную ценность бронирования. Например, отмена дорогого бронирования может быть более чувствительной для бизнеса, чем отмена дешёвого бронирования. Поэтому оценка ожидаемой стоимости помогает платформе приоритизировать работу с клиентами и бронированиями.

После получения предсказанного моделью значения `adr`, мы дополнительно переведем его в ценовой сегмент: low, medium или high. Эти сегменты будут иметь удобный бизнес-формат интерпретации результата для бизнеса и помогут правильнее оценивать и характеризовать предсказанное значение adr.

Польза для компании:
* выделять потенциально более ценные бронирования
* приоритизировать работу с high-value бронированиями
* использовать прогноз стоимости вместе с прогнозом риска отмены
* персонализировать предложения и коммуникацию с клиентами
* эффективнее распределять маркетинговые и операционные ресурсы

### Постановка задач машинного обучения

#### Задача 1: Бинарная классификация отмены бронирования

Первая наша ML задача — бинарная классификация.

Целевая переменная: `adr`.

Посчитаю долю нанов от общего количества объектов для каждого столбца.

```python
def nan_counts(df):
  nans = []
  for c in df.columns:
    nans.append(len(df[pd.isna(df[c])]))

  res = zip(list(df.columns), nans)
  return pd.DataFrame(res, columns = ['column', 'nan_count']).sort_values('nan_count', ascending=False)
```

Такие строки стоит выявить и проверить перед EDA и моделями, потому что они могут искажать анализ и качество моделей.

```python
neg_adr = (df['adr'] < 0).sum()

zero_guests = (df['adults'] + df['children'] + df['babies'] == 0).sum()

zero_nights = (df['stays_in_weekend_nights'] + df['stays_in_week_nights'] == 0).sum()

print('Бронирования, где:', '\n')
print('adr < 0:', neg_adr)
print('0 гостей:', zero_guests)
print('0 ночей:', zero_nights)
```

В данных обнаружены отдельные строки с некорректными или нестандартными значениями, такими как отрицательный adr, бронирования без гостей и бронирования без ночей проживания.

Так как вторая задача проекта связана с прогнозированием adr, отрицательная стоимость проживания является некорректной. Также строки, где заявлено 0 гостей или 0 ночей плохо согласуются с логикой гостиничного бронирования и могут искажать дальнейший анализ.

Поэтому удалим такие строки перед EDA.

```python
print('Размер датасета до удаления строк с некорректными значениями:', df.shape)

df = df[df['adr'] >= 0]
df = df[(df['adults'] + df['children'] + df['babies']) > 0]
df = df[(df['stays_in_weekend_nights'] + df['stays_in_week_nights']) > 0]

print('Размер датасета после удаления строк с некорректными значениями:', df.shape)
```

#### Анализ целевой переменной is_canceled

Целевая переменная `is_canceled`. Она показывает, было ли бронирование отменено или нет и принимает значения:

* `0` - бронирование не отменено
* `1` - бронирование отменено

Так как это целевая переменная для задачи бинарной классификации, важно посмотреть распределение классов. Это поможет нам понять, есть ли сильный дисбаланс между отменёнными и неотменёнными бронированиями.

```python
cancel_counts = df['is_canceled'].value_counts().reset_index()
cancel_counts.columns = ['is_canceled', 'count']
cancel_counts['status'] = cancel_counts['is_canceled'].map({0: 'Не отменено', 1: 'Отменено'})
```

```python
fig = px.pie(cancel_counts,
             names = 'status',
             values = 'count',
             title = 'Доля отменённых и неотменённых бронирований')
fig.update_traces(textinfo = 'value+percent')
fig.show()
```

##### Вывод


Большая часть бронирований относится к классу 0 (отсутствие отмены). Доля неотменённых бронирований составляет примерно 72%, а доля отменённых - около 28%.

Как видно, в нашей задаче классификации есть умеренный дисбаланс классов: отменённых бронирований меньше, чем неотменённых. Не смотря на это, класс 0 достаточно представлен в данных, поэтому задачу можно решать стандартными моделями классификации.

Но дальнейшего моделирования важно учитывать этот дисбаланс. Стоит смотреть не только accuracy, precision и recall, а также F1-score и ROC-AUC, особенно для класса 1, так как именно отменённые бронирования являются важными для нашей бизнес-задачи.

## Feature engineering

Мы хотим создать новые признаки на основе бизнес-логики и результатов проведенного нами EDA и оставить те, которые нам пригодятся далее. Feature Transformation мы объедили с препроцессингом и выделили в отдельный раздел.

Далее мы сравним качество моделей на двух наборах данных: с исходными признаками и с добавленными в ходе feature engeneering и посмотрим качество.

#### Создание копии датасета

Перед созданием новых признаков создадим копию датасета. Это нужно, чтобы сохранить исходную версию данных и отдельно получить версию после feature engineering.

Так как мы будем обучать baseline-модель на исходных признаках и с новыми признаками и хотим посмотреть, улучшиться ли качество.

```python
df_fe = df.copy()
```

### Feature Extraction

#### Признаки длительности проживания

В исходном датасете длительность проживания разбита на две отдельные переменные:
* stays_in_weekend_nights
* stays_in_week_nights

Первая показывает количество ночей в выходные, вторая показывает количество ночей в будние дни.

Для модели может быть полезен общий признак длительности проживания, так как итоговое поведение клиента и стоимость его бронирования могут зависеть от общего количества ночей, а не только от их типа.

Также создадим бинарный признак is_weekend_stay, который будет показывать, включает ли бронирование хотя бы одну ночь в выходные. Такой признак может быть полезен, потому что поездки на выходные и поездки только в будние дни могут отличаться по цели, цене и вероятности отмены.

```python
df_fe['total_nights'] = (df_fe['stays_in_weekend_nights'] + df_fe['stays_in_week_nights'])
df_fe['is_weekend_stay'] = (df_fe['stays_in_weekend_nights'] > 0).astype(int)
```

#### Признаки состава гостей

В нашем датасете состав гостей описан тремя признаками:
* adults
* children
* babies

Для наших моделей может быть полезен общий признак количества гостей total_guests. Он показывает, сколько всего человек было указано в бронировании.

Также создадим признак has_children, который будет показывать, указаны ли в бронировании дети или младенцы или нет. За тем, что семейные бронирования могут отличаться от одиночных или деловых поездок, так как у них могут отличаться стоимости, длительности проживания и другой риск отмены.

Перед этим заполним пропуски в children нулями, потому что если информация о детях отсутствует, в рамках этого простого feature engineering будем считать, что дети не указаны.

```python
#df_fe['children'] = df_fe['children'].fillna(0)

df_fe['total_guests'] = (df_fe['adults'] + df_fe['children'] + df_fe['babies'])
df_fe['has_children'] = ((df_fe['children'] + df_fe['babies']) > 0).astype(int)
```

#### Масштаб бронирования

После создания признаков total_guests и total_nights можно дополнительно оценить общий масштаб бронирования.

Для этого мы создадим признак `booking_intensity`, который будет равен произведению общего количества гостей на общее количество ночей проживания:
* booking_intensity = total_guests * total_nights

Этот признак можно будет интерпретировать, как количество гость-ночей (наподобие трудозатрат - человеко-часов). Например, бронирование на двух гостей и три ночи будет давать значение 6, а бронирование на четырех гостей и пять ночей - значение 20.

Такой признак может быть полезен для моделей, так как он будет одновременно учитывать размер группы идлительность проживания.

```python
df_fe['booking_intensity'] = (df_fe['total_guests'] * df_fe['total_nights'])
```

```python
#df_fe['booking_intensity'].describe()
```

#### Признаки наличия агента и компании

В датасете есть признаки agent и company:
* agent - ID туристического агентства, через которое было сделано бронирование
* company - ID компании или организации, которая сделала бронирование или отвечает за оплату

В этих колонках много промущенных значений, но они не являются обычными пропусками. Они означают, что агент или компания не применимы к конкретному бронированию. Например, если agent отсутствует, это означает, что бронирование пришло не через туристическое агентство.

Поэтому мы создадим два бинарных признака:
* has_agent - было ли бронирование сделано через агента
* has_company - связано ли бронирование с компанией

```python
#df_fe['agent'].unique()
#df_fe['company'].unique()
```

```python
df_fe['has_agent'] = df_fe['agent'].notna().astype(int)
df_fe['has_company'] = df_fe['company'].notna().astype(int)
```

```python
df_fe = df_fe.drop(columns = ['agent', 'company'])
```

#### Группировка стран

Признак country показывает страну клиента. Этот признак может быть полезен для моделей, потому что клиенты из разных стран могут по-разному себя вести.

Однако у признака country достаточно много уникальных значений. Если использовать все страны как отдельные категории, после one-hot encoding появится большое количество новых столбцов. Кроме того, редкие страны будут создавать риск переобучения из-за шума.

Поэтому мы будем группировать страны, а именно оставим только наиболее частые страны, а остальные объединим в категорию `Other`. Это позволит сохранить основную информацию о стране клиента, но сделать признак более компактным для будущего моделирования.

Оставляем только страны, которые встречаются минимум примерно в 1% строк

```python
#df_fe['country'].value_counts().head(20)
```

```python
country = df_fe['country'].value_counts(normalize=True)

top_countries = country[country >= 0.01].index
condition = df_fe['country'].isin(top_countries)

df_fe['country_group'] = np.where(condition, df_fe['country'], 'Other')

df_fe = df_fe.drop('country', axis = 1)
```

```python
df_fe['country_group'].value_counts()
```
